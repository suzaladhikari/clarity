```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ast


# ==============================
# Load Data
# ==============================

combined_data = pd.read_parquet(
    "./datas/combined_data.parquet"
)

xgboost_results = pd.read_json(
    "./modelperformance/xbgoost_predicted.json"
)

garch_results = pd.read_json(
    "./modelperformance/garch_predicted.json"
)

lstm_rnn_results = pd.read_json(
    "./modelperformance/lstm_rnn_predicted.json"
)


# ==============================
# Select Test Data
# ==============================

test_data = combined_data[
    combined_data["date"] >= "2022-01-01"
].reset_index(drop=True)



# ==============================
# Helper Function
# ==============================

def parse_list(x):

    if isinstance(x, list):
        return x

    if isinstance(x, str):
        x = x.strip()

        if not x.startswith("["):
            x = "[" + x

        if not x.endswith("]"):
            x = x + "]"

        return ast.literal_eval(x)

    return list(x)



# ==============================
# Extract XGBoost Predictions
# ==============================

xgb_df = pd.DataFrame(
    {
        "xgboost_predicted": xgboost_results[
            "true_values_xgboost"
        ]
    }
)



# ==============================
# Extract GARCH Predictions
# ==============================

garch_df = pd.DataFrame(
    {
        "garch_predicted": garch_results[
            "true_values_garch"
        ]
    }
)



# ==============================
# Extract LSTM Predictions
# ==============================

lstm_values = lstm_rnn_results.loc[
    lstm_rnn_results["model"] == "lstm",
    "predicted_lstm"
].iloc[0]


lstm_values = parse_list(lstm_values)


lstm_df = pd.DataFrame(
    {
        "lstm_predicted": lstm_values
    }
)



# ==============================
# Extract RNN Predictions
# ==============================

rnn_values = lstm_rnn_results.loc[
    lstm_rnn_results["model"] == "rnn",
    "predicted_rnn"
].iloc[0]


rnn_values = parse_list(rnn_values)


rnn_df = pd.DataFrame(
    {
        "rnn_predicted": rnn_values
    }
)



# ==============================
# Combine Everything
# ==============================

results = pd.concat(
    [
        test_data[
            [
                "date",
                "target_volatility"
            ]
        ],

        xgb_df.reset_index(drop=True),

        lstm_df.reset_index(drop=True),

        rnn_df.reset_index(drop=True),

        garch_df.reset_index(drop=True)

    ],
    axis=1
)



# Remove rows with missing values

results = results.dropna()



# Show data

st.subheader(
    "Actual and Predicted Values"
)

st.dataframe(
    results.head(20),
    use_container_width=True
)



# ==============================
# Scatter Plot
# ==============================

st.subheader(
    "Actual vs Predicted Volatility"
)


fig, ax = plt.subplots(
    figsize=(10,8)
)



sns.scatterplot(
    data=results,
    x="target_volatility",
    y="xgboost_predicted",
    alpha=0.5,
    label="XGBoost",
    ax=ax
)



sns.scatterplot(
    data=results,
    x="target_volatility",
    y="lstm_predicted",
    alpha=0.5,
    label="LSTM",
    ax=ax
)



sns.scatterplot(
    data=results,
    x="target_volatility",
    y="rnn_predicted",
    alpha=0.5,
    label="RNN",
    ax=ax
)



sns.scatterplot(
    data=results,
    x="target_volatility",
    y="garch_predicted",
    alpha=0.5,
    label="GARCH",
    ax=ax
)



# Perfect prediction line

min_value = results[
    "target_volatility"
].min()

max_value = results[
    "target_volatility"
].max()


ax.plot(
    [min_value, max_value],
    [min_value, max_value]
)



ax.set_xlabel(
    "Actual Volatility"
)

ax.set_ylabel(
    "Predicted Volatility"
)


ax.set_title(
    "Actual vs Predicted Volatility Across Models"
)


ax.legend()


plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ast


# =====================================
# Load data
# =====================================

combined_data = pd.read_parquet(
    "./datas/combined_data.parquet"
)

xgboost_results = pd.read_json(
    "./modelperformance/xbgoost_predicted.json"
)

garch_results = pd.read_json(
    "./modelperformance/garch_predicted.json"
)

lstm_rnn_results = pd.read_json(
    "./modelperformance/lstm_rnn_predicted.json"
)



# =====================================
# Test data
# =====================================

test_data = combined_data[
    combined_data["date"] >= "2022-01-01"
].reset_index(drop=True)



# =====================================
# XGBoost predicted values
# =====================================

xgb_df = pd.DataFrame({
    "xgboost_predicted": xgboost_results[
        "true_values_xgboost"
    ].values
})



# =====================================
# GARCH predicted values
# =====================================

garch_df = pd.DataFrame({
    "garch_predicted": garch_results[
        "true_values_garch"
    ].values
})



# =====================================
# Helper for LSTM/RNN lists
# =====================================

def convert_prediction(value):

    if isinstance(value, list):
        return value

    if isinstance(value, str):
        value = value.replace("\n", "")
        return ast.literal_eval(value)

    return list(value)



# =====================================
# LSTM predictions
# =====================================

lstm_values = lstm_rnn_results.loc[
    lstm_rnn_results["model"] == "lstm",
    "predicted_lstm"
].iloc[0]


lstm_values = convert_prediction(lstm_values)


lstm_df = pd.DataFrame({
    "lstm_predicted": lstm_values
})



# =====================================
# RNN predictions
# =====================================

rnn_values = lstm_rnn_results.loc[
    lstm_rnn_results["model"] == "rnn",
    "predicted_rnn"
].iloc[0]


rnn_values = convert_prediction(rnn_values)


rnn_df = pd.DataFrame({
    "rnn_predicted": rnn_values
})



# =====================================
# Combine all models
# =====================================

results = pd.concat(
    [
        test_data[
            [
                "date",
                "target_volatility"
            ]
        ],

        xgb_df,
        garch_df,
        lstm_df,
        rnn_df

    ],
    axis=1
)


# Remove rows where any prediction is missing
results = results.dropna()


print(results.head())
print(results.shape)



# =====================================
# Scatter plot
# =====================================

plt.figure(figsize=(10,8))


sns.scatterplot(
    data=results,
    x="target_volatility",
    y="xgboost_predicted",
    alpha=0.5,
    label="XGBoost"
)


sns.scatterplot(
    data=results,
    x="target_volatility",
    y="lstm_predicted",
    alpha=0.5,
    label="LSTM"
)


sns.scatterplot(
    data=results,
    x="target_volatility",
    y="rnn_predicted",
    alpha=0.5,
    label="RNN"
)


sns.scatterplot(
    data=results,
    x="target_volatility",
    y="garch_predicted",
    alpha=0.5,
    label="GARCH"
)



# Perfect prediction line

min_value = results["target_volatility"].min()
max_value = results["target_volatility"].max()


plt.plot(
    [min_value, max_value],
    [min_value, max_value]
)



plt.xlabel(
    "Actual Volatility"
)

plt.ylabel(
    "Predicted Volatility"
)


plt.title(
    "Actual vs Predicted Volatility Across Models"
)


plt.legend()
plt.grid(True)

plt.show()
```
