```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
from garch import Garch
```

```python
### Lets see how deos garch model work 

data = pd.read_parquet("../updatedtop20datasets/AAPL.parquet")
data.head(5)
```

```python
## Returns column 
first30 = data.head(30)
returnlist = list(first30.price_return)
print(returnlist)
```

```python
## calling the garch model 
garch = Garch(omega=0.0001, alpha= 0.1, beta = 0.85)
variance = garch.computing_variance(returnlist)
variancenext = garch.forecast_next_day(returnlist, variance)
print(f"The volatitily of the next day: {np.sqrt(variancenext)}")
```
