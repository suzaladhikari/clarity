```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('../updatedtop20datasets/AAPL.csv')
data.head(5)
```

```python
### Creating the different rows and columns
### 1. Did the price go up or down today 
def featuers_creation(data):
    data['return'] = data['close'].diff()/ data['close'].shift(1)
    data['return'] = data['return'].fillna(0)
    data['Volatility_7Days'] = data['close'].rolling(window = 7).std()
    data['Volatility_7Days'] = data['Volatility_7Days'].fillna(0)
    data['Volatility_30Days'] = data['close'].rolling(window = 30).std()
    data['Volatility_30Days'] = data['Volatility_30Days'].fillna(0)
    return data
```

### What is rolling Volatility ?!
### Rolling Volatility measures how much a stock's price has been (moving up and down ) over a fixed window of a time. 


```python
featuers_creation(data)
```
