```python
### Libraries to be imported
import pandas as pd 
import numpy as np 
import requests
import os 
```

```python
### Creating datasets for the top 20 companies 
watchlist = {
    "AAPL":  "Apple",
    "MSFT":  "Microsoft", 
    "NVDA":  "Nvidia",
    "GOOGL": "Alphabet (Google)",
    "AMD":  "AMD",
    "META":  "Meta (Facebook)",
    "TSLA":  "Tesla",
    "NFLX": "Netflix",
    "LLY":   "Eli Lilly",
    "AVGO":   "Broadcom",
    "MU":    "Micron Technology",
    "QCOM":   "Qualcomm",
    "UNH":   "UnitedHealth",
    "WMT":   "Walmart",
    "MA":    "Mastercard",
    "JNJ":   "Johnson & Johnson",
    "PG":    "Procter & Gamble",
    "HD":    "Home Depot",
    "ORCL":  "Oracle",
    "JPM": "JPMorgan Chase"
}

print(len(WATCHLIST))
```

```python
def top20_datasets(symbol):
    path = '../updatedtop20datasets/'
    filepath = f'{path}{symbol}.csv'
    if os.path.exists(filepath):
        print(f"Returned from the archive. Didnot do the api call for {symbol}")
        return None
    url = f"https://www.alphavantage.co/query?function=TIME_SERIES_DAILY&symbol={symbol}&apikey=WJGSUHYQRBVMN7T5"
    response = requests.get(url)
    raw = response.json()
    if "Time Series (Daily)" not in raw:
        print(f"Used Enough API Calls. Problem: {symbol} failed: {raw}")
        return None
    time_series = raw["Time Series (Daily)"]
    data = pd.DataFrame.from_dict(time_series, orient = 'index')
    data.columns = ['open', 'high', 'low', 'close', 'volume']
    data.index = pd.to_datetime(data.index)
    data = data.astype(float).sort_index()
    data["Symbol"] = symbol
    data.to_csv(filepath, index=False)
    return data
```

```python
for symbol,name in watchlist.items():
    top20_datasets(symbol)
```
