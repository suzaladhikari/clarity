```python
### Creating a function to create dataset for each of the listed companies 
def company_datasets(symbol):
    path = '../updatedtop20datasets/'
    filepath = f'{path}{symbol}.csv'
    if os.path.exists(filepath):
        return f"Returned from the archive for {symbol}"
    data = yf.download(symbol, start = "1900-01-01")
    data.columns = ["close", "high","low", "open", "volume"]
    data = data.reset_index()
    data["Symbol"] = symbol
    data.to_csv(filepath,index = False)
    return data

    

```
